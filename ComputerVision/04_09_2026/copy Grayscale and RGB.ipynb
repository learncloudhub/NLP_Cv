## Testing with RGB

```python
import cv2 #using cv2 library
```

```python
img_plt= plt.imread('../images/dog.jpg')   # read dog jpg image using matplotlib
```

```python
img_cv = cv2.imread('../images/dog.jpg')   # read dog jpg image using open cv
img_cv
```

```python
b, g, r = cv2.split(img_cv)           
```

```python
b.shape,g.shape, r.shape
```

```python
img_rgb = cv2.merge([b,g,r])   #b,g,r shift r,g,b use merge function
img_gbr = cv2.merge([g,b,r])   # g,b,r
```

```python
plt.subplot(131);
plt.imshow(img_plt)
plt.subplot(132);
plt.imshow(img_rgb)
plt.subplot(133);
plt.imshow(img_gbr)
```
