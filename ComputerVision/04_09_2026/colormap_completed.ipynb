```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Read image as grayscale
img = cv2.imread('../images/dog.jpg', 0)

if img is None:
    raise FileNotFoundError("dog.jpg was not found. Put dog.jpg in the notebook folder.")

plt.figure(figsize=(6, 6))
plt.imshow(img, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.show()
```

```python
# Compare two grayscale colormaps
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='Greys')
plt.title('Greys')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img, cmap='Blues')
plt.title('Blues')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Sequential : Colormap

```python
# Read image from images/dog.jpg
img = cv2.imread('../images/dog.jpg', 0)

if img is None:
    raise FileNotFoundError("images/dog.jpg was not found. Put the image at images/dog.jpg.")

original_map = plt.colormaps['Greys'].reversed()

plt.figure(figsize=(16, 16))

sequential_maps = ['Greys', 'Blues', 'Reds', 'Greens',
                   'Oranges', 'Purples', 'PuRd', 'YlGn']

for i, cmap_name in enumerate(sequential_maps, start=1):
    plt.subplot(4, 4, i)
    plt.imshow(img, cmap=plt.colormaps[cmap_name].reversed())
    plt.title(cmap_name)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## Diverging : Colormap

```python
plt.figure(figsize=(16, 16))

diverging_maps = ['PiYG', 'PRGn', 'BrBG', 'PuOr',
                  'RdGy', 'RdBu', 'Spectral', 'coolwarm']

for i, cmap_name in enumerate(diverging_maps, start=1):
    plt.subplot(4, 4, i)
    plt.imshow(img, cmap=cmap_name)
    plt.title(cmap_name)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## Cyclic : Colormap

```python
plt.figure(figsize=(16, 8))

cyclic_maps = ['hsv', 'twilight', 'twilight_shifted', 'viridis']

for i, cmap_name in enumerate(cyclic_maps, start=1):
    plt.subplot(2, 2, i)
    plt.imshow(img, cmap=cmap_name)
    plt.title(cmap_name)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
