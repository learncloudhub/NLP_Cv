```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('../images/dog.jpg')
```

```python
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
quater_width, quater_height = img.shape[1]//4, img.shape[0]//4
T = np.float32([[1,0,quater_width],[0,1,quater_height]])
T
```

```python
translate_img = cv2.warpAffine(img,T,(img.shape[1],img.shape[0]))
plt.imshow(cv2.cvtColor(translate_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
translate_img = cv2.warpAffine(img,T,(600,600))
plt.imshow(cv2.cvtColor(translate_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
T = np.float32([[1,0,150], [0,1,75]])
translate_img = cv2.warpAffine(img,T,(img.shape[1],img.shape[0]))
plt.imshow(cv2.cvtColor(translate_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
T = np.float32([[1,0,150], [0,1,75]])
translate_img = cv2.warpAffine(img,T,(600,600))
plt.imshow(cv2.cvtColor(translate_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

Image Rotation

```python
img_90 = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
plt.imshow(cv2.cvtColor(img_90, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
img_180 = cv2.rotate(img, cv2.ROTATE_180)
plt.imshow(cv2.cvtColor(img_180, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
center = img.shape[0]//2, img.shape[0]//2
center
```

```python
angle = 30
R = cv2.getRotationMatrix2D(center, angle, 1.0)
rotated_img = cv2.warpAffine(img, R, (400,400))
plt.imshow(cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
angle = 45
R = cv2.getRotationMatrix2D(center, angle, 0.5)
rotated_img = cv2.warpAffine(img, R, (400,400))
plt.imshow(cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
angle = -30
R = cv2.getRotationMatrix2D(center, angle, 0.7)
rotated_img = cv2.warpAffine(img, R, (400,400))
plt.imshow(cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
R = cv2.getRotationMatrix2D((300,300),60,1.0)
rotated_img = cv2.warpAffine(img, R, (600,600), borderValue=(255,255,255))
plt.imshow(cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
