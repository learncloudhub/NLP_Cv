```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
import os
print(os.getcwd ())
```

```python
img = cv2.imread('../images/lena.png', cv2.IMREAD_GRAYSCALE)
plt.imshow(img,cmap='gray')
```

```python
img.shape
```

```python
binary_image = img > 120 
```

```python
plt.imshow(binary_image, cmap='gray')
```

```python
inv_binary_image = img < 120 
```

```python
plt.imshow(inv_binary_image, cmap='gray')
```

```python
binary_image.dtype
```

```python
binary_image = binary_image.astype(np.uint8)  * 255             #binary to black and white pixel value
inv_binary_image = inv_binary_image.astype(np.uint8) * 255
```

```python
binary_image.dtype
```

```python
plt.subplot(121)
plt.imshow(binary_image, cmap='gray')
plt.title("Binary Mask")

plt.subplot(122)
plt.imshow(inv_binary_image, cmap='gray')
plt.title("Inverse Mask")
```

```python
thresh, new_img = cv2.threshold(img, 120, 255, cv2.THRESH_BINARY)

thresh, new_img_inv = cv2.threshold(img, 120, 255, cv2.THRESH_BINARY_INV)
thresh
```

```python
plt.figure(figsize=(16,9))
plt.subplot(141)
plt.title("Original")
plt.imshow(img, cmap = 'gray')
plt.subplot(142)
plt.title("By array operations")
plt.imshow(binary_image, cmap = 'gray')
plt.subplot(143)
plt.title("By functions")
plt.imshow(new_img, cmap = 'gray')
plt.subplot(144)
plt.title("Inverse")
plt.imshow(new_img_inv, cmap = 'gray')
```

#### Otsu Thresholding

```python
thresh, img_otsu = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
```

```python
thresh
```

```python
plt.figure(figsize=(16,9))
plt.subplot(141)
plt.title("Original")
plt.imshow(img, cmap = 'gray')
plt.subplot(142)
plt.title("By array operations")
plt.imshow(binary_image, cmap = 'gray')
plt.subplot(143)
plt.title("By functions")
plt.imshow(new_img, cmap = 'gray')
plt.subplot(144)
plt.title("Otsu")
plt.imshow(img_otsu, cmap = 'gray')
```
