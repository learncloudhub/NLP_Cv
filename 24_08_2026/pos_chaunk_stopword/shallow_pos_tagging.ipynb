```python
# ==========================================
# Part-of-Speech (POS) Tagging
# ==========================================

# Import NLTK
import nltk

# Import word tokenizer
from nltk.tokenize import word_tokenize


# ------------------------------------------
# Download required POS tagging resources
# ------------------------------------------

nltk.download("punkt")                              # used for tokenization     word_tokenize() / sent_tokenize()
nltk.download("punkt_tab")                          # used for Additional data for tokenization
nltk.download("averaged_perceptron_tagger_eng")     # used for pos tagging      nltk.pos_tag()


# ------------------------------------------
# Input Text
# ------------------------------------------

text = "The student studies artificial intelligence."


# ------------------------------------------
# Step 1: Tokenization
# ------------------------------------------

# Divide the sentence into individual tokens
tokens = word_tokenize(text)

print("Tokens:")
print(tokens)


# ------------------------------------------
# Step 2: POS Tagging
# ------------------------------------------

# Assign a POS tag to each token
tags = nltk.pos_tag(tokens)


# ------------------------------------------
# Step 3: Display Word and POS Tag
# ------------------------------------------

print("\nWord and POS Tag:")
print("-----------------")

for word, tag in tags:

    # Display the word and its POS tag
    # :15 reserves 15 spaces for the word
    print(f"{word:15} {tag}")
```

```python
import nltk

from nltk.tokenize import word_tokenize

# work can have different grammatical roles.
sentences = [
    "I work every day.",
    "The work is difficult."
]

for sentence in sentences:

    print("\nSentence:", sentence)

    tokens = word_tokenize(sentence)

    tags = nltk.pos_tag(tokens)

    for word, tag in tags:
        print(word, "->", tag)
```
