```python
# Word Tokenization

import nltk

from nltk.tokenize import word_tokenize

# Punkt is a pre-trained tokenizer model used by NLTK to help determine where sentences and words are separated.
nltk.download('punkt')

# punkt_tab contains additional tokenizer tables/data used by newer versions of NLTK's Punkt tokenizer.
nltk.download('punkt_tab')

text = "I am studying Artificial Intelligence at U.S.A."

# Split text into words/tokens
tokens = word_tokenize(text)

print("Original text:")
print(text)

print("\nTokens:")
print(tokens)
```

```python
# Sentence Tokenization

from nltk.tokenize import sent_tokenize

text = "I am studying NLP. NLP is interesting. I want to learn more."

# Split text into sentences
sentences = sent_tokenize(text)

print("Sentences:")

for sentence in sentences:
    print(sentence)
```

```python
# Word and Sentence Tokenization

from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize

text = "Natural Language Processing is a field of Artificial Intelligence. It allows computers to understand human language."

# Sentence tokenization
sentences = sent_tokenize(text)

print("Number of sentences:", len(sentences))

print("\nSentences:")

for sentence in sentences:
    print(sentence)

# Word tokenization
words = word_tokenize(text)

print("\nNumber of tokens:", len(words))

print("\nTokens:")

for word in words:
    print(word)
```

```python
# ==========================================
# Token Types
# ==========================================
from nltk.tokenize import word_tokenize

text = "I have 10 books."

tokens = word_tokenize(text)

for token in tokens:

    if token.isalpha():
        token_type = "Word"

    elif token.isdigit():
        token_type = "Number"

    elif token.isalnum():
        token_type = "Alphanumeric"

    else:
        token_type = "Punctuation / Symbol"

    print(token, "→", token_type)
```

```python
# ==========================================
# Tokenization Challenges
# ==========================================
from nltk.tokenize import word_tokenize

examples = [
    "I'm happy.",
    "New York is beautiful.",
    "U.S.A. is a country."
]

for text in examples:

    print("\nText:", text)

    tokens = word_tokenize(text)

    print("Tokens:", tokens)
```
