```python
alphabets = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
```

```python
int_to_char = dict(list(enumerate(alphabets)))
```

```python
int_to_char[6]
```

```python
for i, v in enumerate(alphabets):
    print(v,i)
```

```python
char_to_int = dict((v,i) for i, v in enumerate(alphabets))
```

```python
char_to_int['X']
```

```python
### Prepare the dataset
seq_length = 1
x = []
y = []

for i in range(len(alphabets)-seq_length):
    seq_in=alphabets[i]
    seq_out=alphabets[i+1]
    print(seq_in,'--->',seq_out)
    x+=[[char_to_int[seq_in[0]]]]
    y+=[char_to_int[seq_out]]
```

```python
x
```

```python
y
```

```python
import numpy as np
np.shape(x)
```

```python
x=np.reshape(x,(25,1,1)) #sample, time steps/sequence length, feature
```

```python
x
```

```python
x_norm=x/25 #Normalization range 0~25 --> 0.04~1 small scale, nearly same value good in optimization
```

```python
x_norm
```

```python
from tensorflow.keras.utils import to_categorical
```

```python
a = [1,3,2,3,2,1,1,2,3]
```

```python
to_categorical(a)
```

```python
y_new=to_categorical(y)
y_new
```

```python
from keras.layers import SimpleRNN, Dense, Input
from keras.models import Sequential
```

```python
model = Sequential()
```

```python
model.add(Input((1,1)))
model.add(Dense(25,activation='relu'))
model.add(SimpleRNN(32))
model.add(Dense(26,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss = 'categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(x_norm,y_new,epochs=500,batch_size=1)
```

```python
model.evaluate(x_norm,y_new, batch_size=1)
```

```python
from numpy import argmax

new = 'M'
new = char_to_int[new]
new = np.reshape(new,(1,1,1))
new = new/25
pred = model.predict(new, verbose=False)
pred = np.argmax(pred)
int_to_char[pred]
```

### seq_length = 3

```python
### Prepare the dataset
seq_length = 3
x = []
y = []

for i in range(len(alphabets)-seq_length):
    seq_in = alphabets [i:i+seq_length]
    seq_out = alphabets[i+seq_length]
    print(seq_in,'--->',seq_out)
    x.append([char_to_int[char] for char in seq_in])
    y+= [char_to_int[seq_out]]
```

```python
x=np.reshape(x,(23,seq_length,1))
```

```python
x_norm = x/25
```

```python
y_new = to_categorical(y)
```

```python
model = Sequential()
model.add(Input((3,1)))
model.add(Dense(25,activation='relu'))
model.add(SimpleRNN(32))
model.add(Dense(26,activation='softmax'))
model.summary()
```

```python
model.compile(loss = 'categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(x_norm,y_new,epochs=1000,batch_size=1)
```

```python
model.evaluate(x_norm,y_new, batch_size=1)
```

```python
from numpy import argmax

new_c = 'MNO'
new = [char_to_int[char] for char in new_c] 
new = np.reshape(new,(1,3,1))
new = new/25
pred = model.predict(new, verbose=False)
pred = np.argmax(pred)
# out = int_to_char[pred]
print(new_c,'--->',int_to_char[pred])
```
