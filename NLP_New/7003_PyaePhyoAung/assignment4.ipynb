(1) What is ASR?

ASR (Automatic Speech Recognition) is a technology that converts spoken human speech into written text using a computer or AI system.

For example:

Spoken: “Hello, how are you?”
ASR output: Hello, how are you?

ASR systems use techniques such as audio/speech processing, machine learning, and deep learning to recognize words and sentences from an audio signal.

===========================================================

(2) Why can background noise reduce ASR performance?

Background noise can reduce ASR performance because it interferes with the speech signal that the system is trying to recognize.

For example, sounds such as traffic, music, people talking, or wind can make the speaker's words less clear. The ASR model may then confuse sounds or fail to recognize some words correctly.

In simple terms:

Clear speech → easier recognition → higher accuracy

Speech + background noise → harder recognition → lower accuracy

==========================================================

(3) Why is speech recognition harder than simple audio classification?

Speech recognition is harder because it must determine the exact sequence of words being spoken, while simple audio classification usually only needs to identify the type or category of an audio signal.

For example:

Audio classification: Is this sound music, speech, or a dog barking?
Speech recognition: What exactly did the person say? → “I am going to school.”

Speech is also difficult because:

Different speakers have different voices and accents.
Speaking speed can vary.
Background noise can interfere with speech.
Words can sound similar, making them difficult to distinguish.
Speech is sequential, so the order and context of words matter.
Pronunciation can change depending on the speaker and situation.

Therefore, ASR is more complex because it must convert a variable audio signal into an accurate sequence of words, rather than simply assigning an audio clip to a category.
