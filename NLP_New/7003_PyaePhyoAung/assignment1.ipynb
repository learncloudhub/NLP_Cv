```python
print("TensorFlow version: ", tf.__version__)
print("Keras version: ", tf.keras.__version__)
```

```python
from tensorflow.keras.datasets import imdb
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=10000)
```

```python
print("Training Samples: ", len(x_train))
print("Testing Samples: ", len(x_test))
```

```python
print("First Review: ", x_train[0])
print("Second Review: ", x_test[0])
```

```python
print("First Label: ", y_train[0])
print("Second Label: ", y_test[0])
```

```python
print("Length of the first review: ", len(x_train[0]))
print("Length of the second review: ", len(x_test[0]))
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

MAX_LENGTH = 200

x_train_padded = pad_sequences(
    x_train,
    maxlen=MAX_LENGTH,
    padding='post',
    truncating='post'
)

x_test_padded = pad_sequences(
    x_test,
    maxlen=MAX_LENGTH,
    padding='post',
    truncating='post'
)

print("Training data shape:", x_train_padded.shape)
print("Testing data shape:", x_test_padded.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Conv1D, MaxPooling1D
from tensorflow.keras.layers import Flatten, Dense

VOCAB_SIZE = 10000

def create_model(kernel_size):
    model = Sequential([
        Embedding(
            input_dim=VOCAB_SIZE,
            output_dim=128,
            input_length=MAX_LENGTH
        ),

        Conv1D(
            filters=128,
            kernel_size=kernel_size,
            activation='relu'
        ),

        MaxPooling1D(pool_size=2),

        Flatten(),

        Dense(64, activation='relu'),

        Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer='adam',
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model
```

```python
model_2 = create_model(kernel_size=2)

model_2.summary()
```

```python
EPOCHS = 5
BATCH_SIZE = 64

history_2 = model_2.fit(
    x_train_padded,
    y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    verbose=1
)
```

```python
test_loss_2, test_accuracy_2 = model_2.evaluate(
    x_test_padded,
    y_test,
    verbose=1
)

print("Kernel Size = 2")
print("Test Loss:", test_loss_2)
print("Test Accuracy:", test_accuracy_2)
```

```python
model_3 = create_model(kernel_size=3)

model_3.summary()
```

```python
history_3 = model_3.fit(
    x_train_padded,
    y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    verbose=1
)
```

```python
test_loss_3, test_accuracy_3 = model_3.evaluate(
    x_test_padded,
    y_test,
    verbose=1
)

print("Kernel Size = 3")
print("Test Loss:", test_loss_3)
print("Test Accuracy:", test_accuracy_3)
```

```python
model_5 = create_model(kernel_size=5)

model_5.summary()
```

```python
history_5 = model_5.fit(
    x_train_padded,
    y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    verbose=1
)
```

```python
test_loss_5, test_accuracy_5 = model_5.evaluate(
    x_test_padded,
    y_test,
    verbose=1
)

print("Kernel Size = 5")
print("Test Loss:", test_loss_5)
print("Test Accuracy:", test_accuracy_5)
```

```python
import pandas as pd

results = pd.DataFrame({
    "Kernel Size": [2, 3, 5],
    "Test Loss": [test_loss_2, test_loss_3, test_loss_5],
    "Test Accuracy": [test_accuracy_2, test_accuracy_3, test_accuracy_5]
})

results["Test Accuracy (%)"] = results["Test Accuracy"] * 100

results
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

plt.bar(
    results["Kernel Size"].astype(str),
    results["Test Accuracy (%)"]
)

plt.xlabel("Kernel Size")
plt.ylabel("Test Accuracy (%)")
plt.title("Effect of Kernel Size on IMDb Sentiment Classification")

plt.show()
```

```python
best_row = results.loc[results["Test Accuracy"].idxmax()]

print("Best Kernel Size:", best_row["Kernel Size"])
print("Best Test Accuracy:", best_row["Test Accuracy (%)"], "%")
```

(2) Why do we use MaxPooling?

Reduces the size of the feature map.

Keeps the strongest/most important features detected by Conv1D.

Reduces computation and memory usage.

Helps reduce overfitting.

Makes the model more efficient.

Summary:

MaxPooling is used to reduce the feature-map size while keeping the most important features, which makes the CNN faster and helps reduce overfitting.

(3) What happens when the kernel size changes?

Kernel size = 2: Detects short word patterns.

Kernel size = 3: Detects slightly longer patterns and provides a good balance.

Kernel size = 5: Detects longer word patterns and wider context.

A larger kernel generally examines more words at once and can require more computation.

Changing the kernel size can affect the model's accuracy.

Summary:

Changing the kernel size changes the amount of text context the CNN examines. Smaller kernels detect shorter patterns, while larger kernels detect longer patterns. This can affect both the model's accuracy and computation.
