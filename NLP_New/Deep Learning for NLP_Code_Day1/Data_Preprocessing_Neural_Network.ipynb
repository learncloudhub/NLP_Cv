```python
lines = [
    'It was a nice rainy day.',
    'The things were so beautiful in his point.',
    'When your focus is clear, you won.',
    'Many many happy returns of the day.'
]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
tokenizor = Tokenizer()
tokenizor.fit_on_texts(lines)
```

```python
tokenizor.word_docs
```

```python
tokenizor.index_word
```

```python
mat = tokenizor.texts_to_matrix(lines)
mat
```

```python
tokenizor.word_index.keys()
```

```python
seq = tokenizor.texts_to_sequences(lines)
seq
```

```python
padded = pad_sequences(seq, maxlen=10, padding='post')
padded
```

```python
padded = pad_sequences(seq, maxlen=10, padding='pre')
padded
```

```python
n = 0
for sent in seq:
    if n < len(sent):
        n = len(sent)
```

```python
padded = pad_sequences(seq, maxlen=n, padding='post')
padded
```

```python
padded = pad_sequences(seq, maxlen=10, padding='pre')
padded
```
