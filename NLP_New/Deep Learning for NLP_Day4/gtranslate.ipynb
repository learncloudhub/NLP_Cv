```python
from deep_translator import GoogleTranslator

text = "Hello friends! How are you? I will teach you Python Programming"

# English → Marathi
marathi = GoogleTranslator(source="en", target="mr").translate(text)
print("Marathi:", marathi)

# English → Spanish
spanish = GoogleTranslator(source="en", target="es").translate(text)
print("Spanish:", spanish)

# English → Bengali
bengali = GoogleTranslator(source="en", target="bn").translate(text)
print("Bengali:", bengali)
```

```python
myanmar = GoogleTranslator(source="en", target="my").translate(text)
print("Myanmar:", myanmar)
```

```python
text1 = """
Try to understand well
"""

from deep_translator import GoogleTranslator
from requests.exceptions import ConnectionError

try:
    myanmar = GoogleTranslator(source="en", target="my").translate(text1)
    print("Myanmar:", myanmar)
except ConnectionError:
    print("Translation failed: Check your internet connection or DNS settings.")
```
