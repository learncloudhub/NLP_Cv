```python
# ============================================================
# TextRank using Sumy
# ============================================================
# pip install sumy
# ------------------------------------------------------------
# Parser → prepares the document
# Summarizer → analyzes the document
# Summary → final result
# ------------------------------------------------------------


# ------------------------------------------------------------
# 1. Import TextRank
# ------------------------------------------------------------

from sumy.summarizers.text_rank import TextRankSummarizer


# ------------------------------------------------------------
# 2. Import Parser and Tokenizer
# ------------------------------------------------------------

from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer


# ------------------------------------------------------------
# 3. Input text
# ------------------------------------------------------------

text = """
Artificial intelligence is changing education.
AI can help teachers create learning materials.
Students can use AI for personalized learning.
Machine learning is an important part of artificial intelligence.
AI can also provide useful feedback to students.
"""


# ------------------------------------------------------------
# 4. Create Parser
# ------------------------------------------------------------

text_parser = PlaintextParser.from_string(
    text,
    Tokenizer("english")
)


# ------------------------------------------------------------
# 5. Create TextRank Summarizer
# ------------------------------------------------------------

text_rank_summarizer = TextRankSummarizer()


# ------------------------------------------------------------
# 6. Generate Summary
# ------------------------------------------------------------

# Prepared Document
#        ↓
# TextRank Summarizer
#        ↓
# Ranked Sentences
#        ↓
# Top 3 Sentences

summary = text_rank_summarizer(
    text_parser.document,
    sentences_count=3
)


# ------------------------------------------------------------
# 7. Display Summary
# ------------------------------------------------------------


print("TextRank Summary:\n")

for sentence in summary:
    print(sentence)
```
