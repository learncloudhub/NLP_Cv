```python
# Import Library

import random
```

```python
# Create a Small Training Corpus

# text = """
# ChatGPT helps students
# Students learn NLP
# Students learn AI
# Students learn Python
# NLP is interesting
# Artificial Intelligence is powerful
# AI changes the world
# """

text = """
I am PyaePhyoAung and 28 years old. I'm studying AI at UIT. AI Course contains Java, Advanced-python, Maths, Fundamental AI, Machine Learning, Deep Neural Network, Natural language Processing and Computer Vision. Course duration is 6 months. Except of Fundamental AI and Maths, all others subjects have lab exam.
"""
```

```python
# Display the Corpus

print(text)
```

```python
# Split the Text into Words

words = text.split()

print(words)
```

```python
# Build the Markov Dictionary

markov = {}

for i in range(len(words)-1):

    current_word = words[i]
    next_word = words[i+1]

    if current_word not in markov:
        markov[current_word] = []

    markov[current_word].append(next_word)
```

```python
# Display the Dictionary

for word, next_words in markov.items():
    print(word, "→", next_words)
```

```python
# Generate a Sentence

word = "is"

sentence = [word]


for i in range(7):

    if word not in markov:
        break

    word = random.choice(markov[word])

    sentence.append(word)


print(" ".join(sentence))
```

```python
# Choose the Starting Word

start_word = input("Enter starting word: ")

word = start_word

sentence = [word]


for i in range(10):

    if word not in markov:
        break

    word = random.choice(markov[word])

    sentence.append(word)


print("Generated sentence:")
print(" ".join(sentence))
```

```python
# Show Transition Probabilities

for word in markov:

    print("\nCurrent word:", word)

    total = len(markov[word])

    for next_word in set(markov[word]):

        probability = markov[word].count(next_word) / total

        print(
            next_word,
            "Probability:",
            probability
        )
```
