```python
text = """
I am PyaePhyoAung and 28 years old. I'm studying AI at UIT. AI Course contains Java, Advanced-python, Maths, Fundamental AI, Machine Learning, Deep Neural Network, Natural language Processing and Computer Vision. Course duration is 6 months. Except of Fundamental AI and Maths, all others subjects have lab exam.
"""
```

```python
from sumy.summarizers.lex_rank import LexRankSummarizer
from sumy.summarizers.text_rank import TextRankSummarizer
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
```

```python
text_parser = PlaintextParser.from_string(
    text,
    Tokenizer("english")
)
```

```python
lex_rank_summarizer = LexRankSummarizer()
```

```python
summary = lex_rank_summarizer(
    text_parser.document,
    sentences_count=3
)
```

```python
print("LexRank Summary:\n")
for sentence in summary:
    print(sentence)
```

```python
text_rank_summarizer = TextRankSummarizer()
```

```python
summary = text_rank_summarizer(
    text_parser.document,
    sentences_count=3
)
```

```python
print("TextRank Summary:\n")
for sentence in summary:
    print(sentence)
```
