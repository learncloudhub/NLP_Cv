```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
categories = {
    "Sports" : "football basketball match player goal",
    "Technology" : "computer software AI technology smartphone",
    "Food" : "restaurant cooking food recipe delicious"
}
```

```python
new_article = "AI technology software computer"
```

```python
texts = list(categories.values()) + [new_article]
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)
print(X.toarray())
```

```python
print(vectorizer.get_feature_names_out())
```

```python
import pandas as pd
df = pd.DataFrame(
    X.toarray(), columns=vectorizer.get_feature_names_out()
)
df
```

```python
new_vector = vectorizer.transform([new_article])
```

```python
df = pd.DataFrame(
    new_vector.toarray(), columns=vectorizer.get_feature_names_out(),
)
df
```

```python
new_vector = X[-1]
scores = {}

for i, category in enumerate(categories):
    category_vector = X[i]

    matches = ((category_vector > 0).multiply(new_vector > 0)).sum()
    scores[category] = matches

for category, score in scores.items():
    print(f"{category}: {score}")

predicted_category = max(scores, key=scores.get)
print(f"Predicted Category: {predicted_category}")
```
