### 1. Morphological Analysis

```python
words = [
    "retake",
    "dislike",
    "preview",
    "imperfect",
    "impatient",
    "joyful",
    "darkness",
    "slowly",
    "jumped",
    "kindness"
]

prefixes = [
    "re",
    "dis",
    "pre",
    "im"
]

suffixes = [
    "ful",
    "ness",
    "ly",
    "ed",
]

for word in words:
    print("\nWord:", word)

    found_prefixes = None

    for prefixe in prefixes:
        if word.startswith(prefixe):
            found_prefixes = prefixe
            break

    found_suffixes = None

    for suffixe in suffixes:
        if word.startswith(suffixe):
            found_suffixes = suffixe
            break

    print("Prefixe:", found_prefixes)
    print("Suffixe:", found_suffixes)
```

### 2. Tokenization

```python
from nltk import word_tokenize, sent_tokenize
```

```python
text = "Natural Language Processing is interesting. I have 10 books, 2 pens, and a laptop. AI is used in Yangon."

sentences = sent_tokenize(text)
print("Number of sentece:", len(sentences))

for i, sentence in enumerate(sentences, 1):
    print(f"\nSentence {i}: {sentence}")
    tokens = word_tokenize(sentence)
    
    for token in tokens:
        if token.isalpha():
            token_type = "Word"
        elif token.isdigit():
            token_type = "Number"
        else:
            token_type = "Punctuation/Symbol"
            
        print(f"  - {token:<15} : {token_type}")
```

#### 3. POS Tagging

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")

sentences = [
    "The student studies computer science.",
    "She writes Python programs.",
    "The teacher explains artificial intelligence.",
    "Students are learning NLP.",
    "The computer runs quickly."
]

for sentence in sentences:
    tokens = word_tokenize(sentence)
    
    tags = nltk.pos_tag(tokens)
    
    print("\nSentence:", sentence)
    
    for word, tag in tags:
        print(f"{word:15} {tag}")
        
    noun_count = 0
    for word, tag in tags:
        if tag in ["NN", "NNS", "NNP", "NNPS"]:
            noun_count += 1
            
    verb_count = 0
    for word, tag in tags:
        if tag in ["VB", "VBD", "VBG", "VBN", "VBP", "VBZ"]:
            verb_count += 1
            
    print("Number of nouns:", noun_count)
    print("Number of verbs:", verb_count)
```

#### 4. Stopwords Removal

```python
from nltk.corpus import stopwords

nltk.download('stopwords')

text = "The students are learning Python and they are not using a textbook."
original_words = word_tokenize(text)

stop_words = set(stopwords.words('english'))

keep_words = ["not", "are"]

filtered_words = [
    w for w in original_words 
    if w.lower() not in stop_words or w.lower() in keep_words
]

print("Original words:", original_words)
print("Stopwords removed:", [w for w in original_words if w.lower() in stop_words])
print("Final words:", filtered_words)
print("Count before filtering:", len(original_words))
print("Count after filtering:", len(filtered_words))
```

#### 5. Chunking

```python
import nltk
from nltk import word_tokenize, pos_tag

grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>}
    VP: {<VB.*><NP>?}
    PP: {<IN><NP>}
    ADVP: {<RB>+}
"""

sentence = "The smart student works very quickly."

# Tokenization
tokens = word_tokenize(sentence)

# POS Tagging
tagged = pos_tag(tokens)

print("Tokens:")
print(tokens)

print("\nPOS Tags:")
print(tagged)

# Chunking
chunk_parser = nltk.RegexpParser(grammar)
tree = chunk_parser.parse(tagged)

print("\nChunk Tree:")
print(tree)

# Display tree
tree.pretty_print()
```

#### 6. Named Entity Recognition

```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

# Text
text = "Elon Musk visited Google in California on Monday."

# 1. Tokenization
tokens = word_tokenize(text)
print("Tokens:")
print(tokens)

# 2. POS Tagging
tagged = pos_tag(tokens)
print("\nPOS Tags:")
print(tagged)

# 3. Named Entity Recognition
ner_tree = ne_chunk(tagged)

print("\nNER Tree:")
print(ner_tree)

# 4. Display NER tree
ner_tree.pretty_print()
```
