```python
import random

transitions = {
    "Sunny": {"Sunny": 0.7, "Rainy": 0.3},
    "Rainy": {"Sunny": 0.4, "Rainy": 0.6}
}

weather = input("Enter today's weather (Sunny/Rainy): ").strip().capitalize()

if weather not in transitions:
    print("Please enter either Sunny or Rainy.")
else:
    for day in range(1, 4):
        next_weather = random.choices(
            list(transitions[weather].keys()),
            weights=list(transitions[weather].values())
        )[0]

        print(f"Day {day}: {next_weather}")

        weather = next_weather
```
