```python
import nltk
from nltk.corpus import wordnet
from nltk.tokenize import word_tokenize

nltk.download("wordnet")
nltk.download("punkt")

CONTEXT_RULES = {
    "bank": {
        "financial": [
            "money",
            "deposit",
            "deposited",
            "withdraw",
            "loan",
            "account",
            "cash",
            "finance",
        ],
        "river": ["river", "water", "shore", "fishing", "stream"],
    },
    "bat": {
        "animal": ["fly", "flew", "cave", "wing", "animal", "night"],
        "sports": ["ball", "baseball", "cricket", "hit", "player", "game"],
    },
}

WORDNET_SYNSETS = {
    "bank": {
        "financial": "bank.n.02",
        "river": "bank.n.01",
    },
    "bat": {
        "animal": "bat.n.01",
        "sports": "bat.n.02",
    },
}


def rule_based_wsd(sentence, target_word):
    target_word_lower = target_word.lower()

    words = [word.lower() for word in word_tokenize(sentence)]

    print("=" * 60)
    print(f"Sentence: \"{sentence}\"")
    print(f"Target Word: '{target_word}'")
    print("-" * 60)

    print("Possible WordNet Synsets & Definitions:")
    all_synsets = wordnet.synsets(target_word_lower, pos=wordnet.NOUN)
    for syn in all_synsets[:4]:
        print(f" - {syn.name()}: {syn.definition()}")
    print()

    target_rules = CONTEXT_RULES.get(target_word_lower, {})

    scores = {}
    matches = {}

    for sense, keywords in target_rules.items():

        matching_words = [word for word in words if word in keywords]
        scores[sense] = len(matching_words)
        matches[sense] = matching_words

    best_sense = max(scores, key=scores.get)
    best_score = scores[best_sense]
    matched_keywords = matches[best_sense]

    synset_name = WORDNET_SYNSETS[target_word_lower].get(best_sense)
    synset_obj = wordnet.synset(synset_name)
    definition = synset_obj.definition()

    print("Disambiguation Result:")
    print(f" - Selected Sense: {best_sense.upper()}")
    print(f" - WordNet Synset: {synset_name}")
    print(f" - Definition: {definition}")
    print(f" - Matching Keywords: {matched_keywords}")
    print(f" - Score: {best_score}")
    print("=" * 60 + "\n")

test_cases = [
    ("I deposited money in the bank.", "bank"),
    ("We sat on the bank of the river.", "bank"),
    ("The bat flew out of the cave.", "bat"),
    ("He hit the ball with a bat.", "bat"),
]

for sentence, target in test_cases:
    rule_based_wsd(sentence, target)
```
