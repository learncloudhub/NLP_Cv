```python
# Morphological Analysis
# Analyze prefixes and suffixes in words
# simple rule-based morphological analyzer

words = [
    "retake",
    "dislike",
    "preview",
    "imperfect",
    "impatient",
    "joyful",
    "darkness",
    "slowly",
    "jumped",
    "kindness"
]

prefixes = [
    "un",
    "re",
    "dis",
    "pre",
    "im",
    "in"
]

suffixes = [
    "ing",
    "ed",
    "ness",
    "er",
    "less",
    "ful",
    "ly"
]

# Analyze each word
for word in words:

    print("\nWord:", word)

    # Check prefix
    found_prefix = None

    for prefix in prefixes:
        if word.startswith(prefix):
            found_prefix = prefix
            break

    # Check suffix
    found_suffix = None

    for suffix in suffixes:
        if word.endswith(suffix):
            found_suffix = suffix
            break

    # Display result
    print("Prefix:", found_prefix)
    print("Suffix:", found_suffix)
```

```python
# Morphological Analysis
# Find prefix, root, and suffix
# simple rule-based morphological analyzer

words = [
    "unhappy",
    "happiness",
    "playing",
    "teacher",
    "careless",
    "rewrite"
]

prefixes = ["un", "re", "dis", "pre", "im", "in"]

suffixes = ["ing", "ed", "ness", "er", "less", "ful", "ly"]


for word in words:

    prefix = ""
    suffix = ""
    root = word

    # Find prefix
    for p in prefixes:
        if word.startswith(p):
            prefix = p
            root = word[len(p):]
            break

    # Find suffix
    for s in suffixes:
        if root.endswith(s):
            suffix = s
            root = root[:-len(s)]
            break

    print("\nWord :", word)
    print("Prefix:", prefix)
    print("Root  :", root)
    print("Suffix:", suffix)
```
