```python
# No new dataset required

from nltk.stem import PorterStemmer

# Create stemmer
stemmer = PorterStemmer()

words = [
    "play",
    "playing",
    "played",
    "plays",
    "player",   # This is one of the reasons lemmatization is useful. It uses more linguistic information to determine the base/dictionary form.
    "connect",
    "connected",
    "connecting",
    "connection",
    "connections",
    "connectivity",
    "study",   # a stemmer is an algorithm for suffix reduction, not a complete morphological analyzer
    "studies",
    "studying",
    "studied"
]



# Apply stemming
for word in words:

    stem = stemmer.stem(word)

    print(f"{word:15} -> {stem}")
```
