```python
import nltk

from nltk.corpus import stopwords

# Download stopwords
nltk.download('stopwords')

# Example text
text = "The student is studying artificial intelligence."

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Split text into words
words = text.split()

# Remove stopwords
filtered_words = []

# word.isalpha()   checks whether the token contains only alphabetic characters.

for word in words:

    if word.lower() not in stop_words and word.isalpha():
        filtered_words.append(word)

print("Original:")
print(words)

print("\nAfter stopword removal:")
print(filtered_words)
```

```python
# ==========================================
# Stopword Removal with Negation Words for Sentiment Analysis
# ==========================================

# Import NLTK
import nltk

# Import English stopword list
from nltk.corpus import stopwords


# ------------------------------------------
# Download English stopwords
# ------------------------------------------

nltk.download("stopwords")


# ------------------------------------------
# Example text
# ------------------------------------------

text = "I do not like this movie."


# ------------------------------------------
# Get English stopwords
# ------------------------------------------

stop_words = set(stopwords.words("english"))


# ------------------------------------------
# Define important negation words
# ------------------------------------------

# Keep these words because they can change
# the meaning of a sentence.

negation_words = {
    "not",
    "no",
    "never",
    "neither",
    "nor"
}


# ------------------------------------------
# Remove negation words from the stopword list
# ------------------------------------------

stop_words -= negation_words


# ------------------------------------------
# Split text into words
# ------------------------------------------

words = text.split()


# ------------------------------------------
# Remove stopwords
# ------------------------------------------

filtered_words = []

for word in words:

    # Keep the word if it is not a stopword
    if word.lower() not in stop_words:
        filtered_words.append(word)


# ------------------------------------------
# Display results
# ------------------------------------------

print("Original:", words)

print("After stopword removal:", filtered_words)
```
