```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk
from nltk.tree import Tree

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

# Dataset
sentences = [
    "Elon Musk visited Singapore in January 2026.",
    "Microsoft opened a new office in Yangon.",
    "Dr. Aung Min presented the project at the University of Information Technology.",
    "The company invested 5 million dollars in Myanmar.",
    "Google organized an AI conference in Bangkok."
]

# Dictionary to store entities by category
entities = {
    "PERSON": [],
    "ORGANIZATION": [],
    "GPE": [],
    "DATE": [],
    "MONEY": []
}

# Process each sentence
for sentence in sentences:

    # 1. Tokenize
    tokens = word_tokenize(sentence)

    # 2. POS tagging
    tagged_tokens = pos_tag(tokens)

    # 3. Named Entity Recognition
    tree = ne_chunk(tagged_tokens)

    # 4. Extract recognized entities
    for subtree in tree:
        if isinstance(subtree, Tree):
            entity_name = " ".join(word for word, tag in subtree.leaves())
            entity_type = subtree.label()

            # 5. Group entities into categories
            if entity_type in entities:
                if entity_name not in entities[entity_type]:
                    entities[entity_type].append(entity_name)

# 6. Display extracted entities and categories
print("========== Named Entity Recognition ==========\n")

for category, entity_list in entities.items():
    print(category + ":")

    for entity in entity_list:
        print(entity)

    print()

# 7. Count entities in each category
print("========== Entity Counts ==========\n")

for category, entity_list in entities.items():
    print(f"{category}: {len(entity_list)}")
```

```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

# Download required NLTK resource
nltk.download('punkt')
nltk.download('punkt_tab')

# Paragraph
paragraph = """
Modern computers are becoming increasingly powerful.

Developers develop applications for users who are using different devices.

The applications process information and provide useful services.

Students are learning programming techniques and practicing their skills.
"""

# Create Porter Stemmer
stemmer = PorterStemmer()

# Tokenize the paragraph
words = word_tokenize(paragraph)

# Remove punctuation
words = [
    word for word in words
    if word not in string.punctuation
]

# Count changed words
changed_words = 0

# Display header
print(f"{'Original':<15}{'Stem'}")
print("-" * 30)

# Apply stemming
for word in words:
    stem = stemmer.stem(word)

    print(f"{word:<15}{stem}")

    # Count words changed by stemming
    if word != stem:
        changed_words += 1

# Display total
print("\nTotal number of changed words:", changed_words)
```
