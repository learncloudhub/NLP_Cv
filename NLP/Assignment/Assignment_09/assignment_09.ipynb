```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents = [
	"Python programming language is useful",
	"Python programming language is popular",
	"Football players practice daily",
	"Football players practice professionally"
]
```

```python
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(documents)
```

```python
print(tfidf.get_feature_names_out())
```

```python
print(X.toarray())
```

```python
import pandas as pd
df = pd.DataFrame(X.toarray(), columns=tfidf.get_feature_names_out())
df
```

```python
vectorizer = TfidfVectorizer()
matrix = vectorizer.fit_transform(documents)
words = vectorizer.get_feature_names_out()
for i in range(len(documents)):
    values = matrix[i].toarray()[0]
    highest_index = values.argmax()
    important_word = words[highest_index]

    # 3. Print the most important word
    print(f"Document {i+1}: {important_word}")
```
