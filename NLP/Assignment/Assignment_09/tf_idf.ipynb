```python
from numpy.ma.core import argmax
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents = [
	"Python programming language is useful",
	"Python programming language is popular",
	"Football players practice daily",
	"Football players practice professionally"
]
```

```python
vectorizer = TfidfVectorizer()
matrix = vectorizer.fit_transform(documents)
```

```python
words = vectorizer.get_feature_names_out()
```

```python
for i in range(len(documents)):
    values = matrix[i].toarray()[0]
    highest_index = values.argmax()
    important_word = words[highest_index]
    print(f'{i+1} : {highest_index} : {words[highest_index]}')
```
