```python
from collections import Counter, defaultdict
import nltk
from nltk.tokenize import word_tokenize
from nltk.util import trigrams


text = """
I love natural language processing.
I love machine learning.
I study natural language processing.
I study machine learning.
"""

tokens = word_tokenize(text)
trigram_list = list(trigrams(tokens))

trigram_counts = Counter(trigram_list)
context_counts = defaultdict(int)

for w1, w2, w3 in trigram_list:
    context_counts[(w1, w2)] += 1

current_words = ("I", "love")
context_total = context_counts[current_words]

next_word_probs = {}

if context_total > 0:
    for (w1, w2, w3), count in trigram_counts.items():
        if (w1, w2) == current_words:
            prob = count / context_total
            next_word_probs[w3] = prob

print("========== TRIGRAM NEXT-WORD PREDICTION ==========\n")
print(f"Current Context: {current_words}\n")

print("Possible Next Words and Probabilities:")
for word, prob in next_word_probs.items():
    print(f"  Word: '{word}' -> Probability: {prob:.2f}")

max_prob = max(next_word_probs.values())
most_likely = [
    word for word, prob in next_word_probs.items() if prob == max_prob
]

print(f"\nMost Likely Next Word(s): {', '.join(most_likely)} ({max_prob:.2f})")
```
