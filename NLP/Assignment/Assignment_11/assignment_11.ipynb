```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Create dataset
data = {
    "text": [
        "The phone is excellent",
        "I love this phone",
        "The camera quality is amazing",
        "This laptop works perfectly",
        "The phone is terrible",
        "I hate this phone",
        "The camera quality is poor",
        "This laptop is very slow"
    ],
    "label": [
        "positive",
        "positive",
        "positive",
        "positive",
        "negative",
        "negative",
        "negative",
        "negative"
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)
print()


# 2. Convert text to TF-IDF
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df["text"])

vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)
print()


# 3. Separate positive and negative documents
# Convert Pandas Series to NumPy array using .to_numpy()

positive_mask = (df["label"] == "positive").to_numpy()
negative_mask = (df["label"] == "negative").to_numpy()

positive_X = X[positive_mask]
negative_X = X[negative_mask]

print("Positive TF-IDF shape:", positive_X.shape)
print("Negative TF-IDF shape:", negative_X.shape)
print()


# 4. Calculate positive centroid
positive_centroid = positive_X.mean(axis=0)

# 5. Calculate negative centroid
negative_centroid = negative_X.mean(axis=0)


# Convert sparse/matrix result to NumPy array
positive_centroid = positive_centroid.A
negative_centroid = negative_centroid.A


# 6. Display centroids
print("Positive Centroid:")
print(positive_centroid)
print()

print("Negative Centroid:")
print(negative_centroid)
print()


# Display centroid values with vocabulary
centroid_table = pd.DataFrame({
    "word": vocabulary,
    "positive_centroid": positive_centroid.flatten(),
    "negative_centroid": negative_centroid.flatten()
})

print("Centroid values:")
print(centroid_table)
print()


# 7. New sentence
new_text = ["I love this phone"]


# 8. Convert new sentence to TF-IDF
new_vector = vectorizer.transform(new_text)

print("New sentence:")
print(new_text[0])
print()

print("New sentence TF-IDF vector:")
print(new_vector.toarray())
print()


# 9. Calculate cosine similarities
positive_similarity = cosine_similarity(
    new_vector,
    positive_centroid
)[0][0]

negative_similarity = cosine_similarity(
    new_vector,
    negative_centroid
)[0][0]


# 10. Display results
print("Positive similarity:", positive_similarity)
print("Negative similarity:", negative_similarity)


if positive_similarity > negative_similarity:
    prediction = "positive"
else:
    prediction = "negative"

print("Final prediction:", prediction)
```
