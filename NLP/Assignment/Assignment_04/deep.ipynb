```python
import nltk
from nltk import CFG
from nltk.parse import ChartParser
```

```python
# Define CFG grammar
grammar = CFG.fromstring("""
S -> NP VP
NP -> Det Adj N
NP -> Det N

VP -> V NP
VP -> V NP PP ADVP


PP -> P NP

ADVP -> Adv

Det -> 'the' | 'a'
Adj -> 'clever'
N -> 'teacher' | 'book' | 'library'
V -> 'reads'
P -> 'in'
Adv -> 'today'
""")

parser = ChartParser(grammar)
sentence = "the clever teacher reads a book in the library today"
words = sentence.split()
trees = parser.parse(words)
for tree in trees:
    print(tree)
    tree.pretty_print()
```
