```python
from nltk import word_tokenize
```

```python
text = """
Microsoft developed a new artificial intelligence system
for students in Yangon. The system helps students learn
Python programming.
"""
```

#### 1) Tokenize the text

```python
tokens = word_tokenize(text)
print("Tokens: ", tokens)
```

#### 2) Display the keywords

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
keywords = []
for word in tokens:
    if word.isalpha() and word.lower() not in stop_words:
        if word not in keywords: keywords.append(word)
for word in keywords:
    print(word)
```

#### 3) Name Entities

```python
from nltk import pos_tag, ne_chunk
pos_name = pos_tag(tokens)
chunk = ne_chunk(pos_name)
for tree in chunk:
    if hasattr(tree, 'label'):
        entity_name = " ".join(word for word, tag in tree.leaves())
        if tree.label() == "ORGANIZATION":
            entity_type = "Organization"
        elif tree.label() in ("GPE", "LOCATION"):
            entity_type = "Location"
        elif tree.label() == "PERSON":
            entity_type = "Person"
        else:
            entity_type = tree.label()
        print(f'{entity_type}: {entity_name}')
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
for ent in doc.ents:
    print(f"{ent.label_}: {ent.text}")
```

#### 4) MExtract and display Noun Phrase

```python
import nltk
from nltk import pos_tag, word_tokenize

text = """
Microsoft developed a new artificial intelligence system
for students in Yangon. The system helps students learn
Python programming.
"""

tokens = word_tokenize(text)
pos_tagged = pos_tag(tokens)

grammar = r"""
    NP: {<JJ><NN><NN>}     # artificial intelligence system
        {<NNS>}           # students
        {<NNP><NN>}        # Python programming (if programming is tagged NN)
        {<NNP>}           # Microsoft, Yangon, Python
"""

chunk_parser = nltk.RegexpParser(grammar)
tree = chunk_parser.parse(pos_tagged)

print("Noun Phrases:")
for subtree in tree.subtrees():
    if subtree.label() == "NP":
        phrase = " ".join(word for word, tag in subtree.leaves())
        print(phrase)

print()
```

#### 5) and 6). One Synonym and Display

```python
import nltk
from nltk.corpus import wordnet


def get_one_synonym(target_word, pos=None):
    wn_pos = None
    if pos == "V":
        wn_pos = wordnet.VERB

    synsets = wordnet.synsets(target_word, pos=wn_pos)

    for synset in synsets:
        for lemma in synset.lemmas():
            syn = lemma.name().replace("_", " ")
            if (
                syn.lower() != target_word.lower()
                and not target_word.lower().startswith(syn.lower())
                and syn.lower() not in ["larn"]
            ):
                return syn
    return "None"

syn_developed = get_one_synonym("developed", pos="V")
syn_learn = get_one_synonym("learn", pos="V")

print(f"developed → {syn_developed}")
print(f"learn → {syn_learn}")
```
