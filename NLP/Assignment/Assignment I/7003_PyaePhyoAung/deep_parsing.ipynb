```python
import nltk
from nltk import CFG
from nltk.parse import ChartParser
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

grammar = CFG.fromstring(
    """
    S -> NP VP
    NP -> Det N
    VP -> V NP
    Det -> 'The' | 'a' | 'the'
    N -> 'student' | 'book'
    V -> 'reads'
"""
)

parser = ChartParser(grammar)

sentences = ["The student reads a book.", "The student book reads a."]

for sentence in sentences:
    tokens = [
        word for word in word_tokenize(sentence) if word not in [".", "?", "!"]
    ]

    print(f"Sentence: {sentence}")

    parses = list(parser.parse(tokens))

    if parses:
        print("Result: Valid sentence")
        print("Parse tree: Generated")
        for tree in parses:
            tree.pretty_print()
    else:
        print("Result: Invalid sentence")
        print("No parse tree found")

    print()
```
