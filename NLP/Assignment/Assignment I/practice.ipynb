```python
input_text = """Microsoft developed a new artificial intelligence system
for students in Yangon. The system helps students learn
Python programming."""
```

1. Tokenize the text

```python
from nltk import word_tokenize
tokens = word_tokenize(input_text)
print(tokens)
```

2. Display the keywords

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
print(stop_words)
```

```python
keywords = []
for word in tokens:
    if word.isalpha() and word.lower() not in stop_words:
        if word not in keywords: keywords.append(word)
print(keywords)
```

3. Identify and Display the Name Entity

```python
from nltk import pos_tag, ne_chunk
pos_name = pos_tag(tokens)
chunk = ne_chunk(pos_name)
```

```python
for tree in chunk:
    if hasattr(tree, 'label'):
        entity_name = " ".join(word for word, tag in tree.leaves())
        if tree.label() == "ORGANIZATION":
            entity_type = "Organization"
        elif tree.label() in ("GPE", "LOCATION"):
            entity_type = "Location"
        elif tree.label() == "PERSON":
            entity_type = "Person"
        else:
            entity_type = tree.label()
        print(f"{entity_type}: {entity_name}")
```

```python
import spacy
nlp = spacy.load('en_core_web_sm')
doc = nlp(input_text)
for ent in doc.ents:
    print(f"{ent.label_}: {ent.text}")
```
