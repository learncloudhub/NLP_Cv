```python
from nltk import bigrams, FreqDist
from nltk.util import ngrams
from collections import Counter

from transformers.utils.type_validators import probability
```

```python
text = "I love natural language processing I love machine learning"
```

```python
words = text.lower().split()
print(words)
```

```python
bigrams = list(ngrams(words, 2))
print(bigrams)
trigrams = list(ngrams(words, 3))
print(trigrams)
# fdist = FreqDist(words)
# print(fdist)
```

```python
bigrams_count = Counter(bigrams)
print(bigrams_count)
```

```python
trigrams_count = Counter(trigrams)
print(trigrams_count)
```

```python
word_count = Counter(words)
print(word_count)
```

```python
for trigram, count in trigrams_count.items():
    first_two_words = (trigram[0], trigram[1])
    
    print("\ntrigram = ", trigram)
    print("first_two_words = ", first_two_words)
    
    print("count of trigram = ", count)
    print("count of first two words = ", bigrams_count[first_two_words])

    probability = count / bigrams_count[first_two_words]
    print("probability =", count, "/", bigrams_count[first_two_words])

    print(trigram, probability)
    print("\n--------------------")
    
```
