```python
import nltk
import random
import re
from nltk.corpus import wordnet
```

```python
sentence = input("Please enter a sentence: ")
```

```python
word = input("Please enter a word to replace: ")
```

```python
synsets = wordnet.synsets(word)
synonyms = set()
for synset in synsets:
    for lemma in synset.lemmas():
        synonyms.add(lemma.name().replace('_', ''))
```

```python
synonyms.discard(word.lower())
```

```python
if len(synonyms) > 0:
    selected_synonym = random.choice(list(synonyms))
    new_sentence = re.sub(
        r"\b" + re.escape(word) + r"\b",
        selected_synonym,
        sentence,
        count=1,
        flags=re.IGNORECASE
    )
    print(f"Original sentence: {sentence}")
    print(f"Selected synonyms: {selected_synonym}")
    print(f"Replaced sentence: {new_sentence}")
else:
    print(f"No synonyms found for {word}")
```
