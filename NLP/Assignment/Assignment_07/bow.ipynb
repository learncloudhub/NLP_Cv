```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
categories = {
    "Sports": "football basketball match player goal",
    "Technology": "computer software AI technology smartphone",
    "Food": "restaurant cooking food recipe delicious"
}
```

```python
texts = list(categories.values())
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)
print(X.toarray())
```

```python
import pandas as pd
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
df
```

```python
new_vector = X[-1]
```

```python
new_article = "Messi is the best footballer. He likes food. He uses AI. His match are amazing"
```

```python
new_vector = vectorizer.transform([new_article])
new_vector.toarray()
```
