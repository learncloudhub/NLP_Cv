```python
import nltk
from nltk.util import ngrams
from collections import Counter
```

```python
text = """
I love NLP
I love AI
"""
```

```python
words = text.lower().split()

print(words)
```

```python
unigrams = list(ngrams(words,1))

print(unigrams)
```

```python
bigrams = list(ngrams(words,2))

print(bigrams)
```

```python
trigrams = list(ngrams(words,3))

print(trigrams)
```

```python
unigram_count = Counter(unigrams)

print(unigram_count)
```

```python
bigram_count = Counter(bigrams)

print(bigram_count)
```

```python
word_count = Counter(words)

for bigram, count in bigram_count.items():

    first_word = bigram[0]

    print("bigram =", bigram)
    print("first_word =", first_word)

    print("count of bigram =", count)
    print("count of first word =", word_count[first_word])

    probability = count / word_count[first_word]

    print("probability =", count, "/", word_count[first_word])

    print(bigram, probability)
    print("\n--------------------")
```

```python
current_word = "love"

possible_words = []

for bigram, count in bigram_count.items():

    if bigram[0] == current_word:

        possible_words.append(bigram[1])

print(possible_words)
```
