```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = [
    "I love NLP",
    "I love AI",
    "NLP is interesting",
    "AI is powerful"
]

print(documents)
```

```python
tfidf = TfidfVectorizer()

X = tfidf.fit_transform(documents)
```

```python
print(tfidf.get_feature_names_out())
```

```python
print(X.toarray())
```

```python
import pandas as pd

tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=tfidf.get_feature_names_out()
)

tfidf_df
```

```python
new_text = [
    "I love AI"
]

new_vector = tfidf.transform(new_text)

print(new_vector.toarray())
```

```python

# another example
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfTransformer

# ၁။ နမူနာ စာသားများ (Customer Reviews)
reviews = [
    "The phone battery is bad",
    "The phone screen is good"
]

# ၂။ အဆင့် (၁) - Bag of Words (Count Vectorizer) အရင်လုပ်ခြင်း
count_vectorizer = CountVectorizer()
word_counts = count_vectorizer.fit_transform(reviews)

# ၃။ အဆင့် (၂) - ထွက်လာသော Count ကိုမှ TF-IDF Transformer ဖြင့် ပြောင်းလဲခြင်း
tfidf_transformer = TfidfTransformer()
tfidf_matrix = tfidf_transformer.fit_transform(word_counts)

# ၄။ ရလဒ်ကို ဇယားကွက်ဖြင့် လှလှပပ ပြသခြင်း
feature_names = count_vectorizer.get_feature_names_out()
df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)
print(df)
```
