```python
# ============================================================
# Step 1: Import Libraries
# ============================================================

import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# ============================================================
# Step 2: Create Dataset
# ============================================================

data = {
    "text": [
        "I love this movie",
        "This movie is amazing",
        "I like this product",
        "This is excellent",
        "I hate this movie",
        "This movie is terrible",
        "I dislike this product",
        "This is bad"
    ],

    "label": [
        "positive",
        "positive",
        "positive",
        "positive",
        "negative",
        "negative",
        "negative",
        "negative"
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)
```

```python
# ============================================================
# Step 3: Convert Text to TF-IDF
# ============================================================

tfidf = TfidfVectorizer()

X = tfidf.fit_transform(df["text"])

print("\nTF-IDF Shape:")
print(X.shape)
```

```python
# ============================================================
# Step 4: Convert TF-IDF to Array
# ============================================================

X_array = X.toarray()

print("\nTF-IDF Array:")
print(X_array)
```

```python
# ============================================================
# Step 5: Separate Classes
# ============================================================

positive_data = X_array[df["label"] == "positive"]  # Select TF-IDF vectors of positive documents
negative_data = X_array[df["label"] == "negative"]  # Select TF-IDF vectors of negative documents

print("\nPositive Documents:")
print(positive_data)

print("\nNegative Documents:")
print(negative_data)
```

```python
# ============================================================
# Step 6: Calculate Centroids
# ============================================================

# Calculate the average TF-IDF value for each word across all positive documents.
positive_centroid = positive_data.mean(
    axis=0
).reshape(1, -1)   # Change to 2D for cosine similarity

negative_centroid = negative_data.mean(
    axis=0
).reshape(1, -1)

print("\nPositive Centroid:")
print(positive_centroid)

print("\nNegative Centroid:")
print(negative_centroid)
```

```python
# ============================================================
# Step 7: Show Vocabulary
# ============================================================

features = tfidf.get_feature_names_out()

print("\nVocabulary:")

for word in features:
    print(word)
```

```python
print("\nPositive Centroid Values:")

for word, value in zip(
    features,
    positive_centroid[0]
):
    print(word, "=", value)
```

```python
# ============================================================
# Step 8: New Text
# ============================================================

new_text = [
    "I love this movie"
]

new_vector = tfidf.transform(new_text)

print("\nNew Text:")
print(new_text)
```

```python
# ============================================================
# Step 9: Calculate Similarity
# ============================================================

positive_similarity = cosine_similarity(
    new_vector,
    positive_centroid
)[0][0]

negative_similarity = cosine_similarity(
    new_vector,
    negative_centroid
)[0][0]

print("\nPositive Similarity:")
print(positive_similarity)

print("\nNegative Similarity:")
print(negative_similarity)
```

```python
# ============================================================
# Step 10: Classification
# ============================================================

if positive_similarity > negative_similarity:
    prediction = "positive"
else:
    prediction = "negative"

print("\nPrediction:")
print(prediction)
```

```python
'''
Dataset
   ↓
Text + Labels
   ↓
TF-IDF
   ↓
TF-IDF Vectors
   ↓
Separate Positive / Negative
   ↓
Positive Centroid + Negative Centroid
   ↓
New Text
   ↓
TF-IDF
   ↓
Cosine Similarity
   ↓
Higher Similarity
   ↓
Prediction
'''
```
