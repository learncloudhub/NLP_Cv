```python
# ============================================================
# LexRank Summarization using Sumy
# ============================================================

# LexRank is an EXTRACTIVE summarization technique.
#
# It ranks sentences based on their similarity
# and importance within the document.
#
# Basic Flow:
#
# Text
#   ↓
# Parser
#   ↓
# LexRank Summarizer
#   ↓
# Sentence Similarity
#   ↓
# Sentence Ranking
#   ↓
# Select Important Sentences
#   ↓
# Summary


# ============================================================
# Step 1: Import LexRank
# ============================================================

# Import LexRankSummarizer from the Sumy library.
# This tool is used to rank sentences and
# select important sentences.

from sumy.summarizers.lex_rank import LexRankSummarizer
```

```python
# ============================================================
# Step 2: Import Parser and Tokenizer
# ============================================================

# PlaintextParser prepares the input text
# for the summarization process.

from sumy.parsers.plaintext import PlaintextParser

# Tokenizer divides the text into sentences and words.

from sumy.nlp.tokenizers import Tokenizer
```

```python
# ============================================================
# Step 3: Create Input Text
# ============================================================

# We use the SAME text as the TextRank practical
# so that we can compare TextRank and LexRank.

text = """
Artificial intelligence is changing education.
AI can help teachers create learning materials.
Students can use AI for personalized learning.
Machine learning is an important part of artificial intelligence.
AI can also provide useful feedback to students.
"""
```

```python
# ============================================================
# Step 4: Create Parser
# ============================================================

# Formula:
#
# Text
#   ↓
# Parser
#   ↓
# Prepared Document

text_parser = PlaintextParser.from_string(
    text,
    Tokenizer("english")
)
```

```python
# ============================================================
# Step 5: Create LexRank Summarizer
# ============================================================

# Create a LexRank summarizer.
#
# LexRank analyzes relationships/similarity between
# sentences and ranks their importance.

lex_rank_summarizer = LexRankSummarizer()
```

```python
# ============================================================
# Step 6: Generate Summary
# ============================================================

# Send the prepared document to LexRank.
#
# Formula:
#
# Prepared Document
#        ↓
# LexRank
#        ↓
# Rank Sentences
#        ↓
# Select Top 3 Sentences

summary = lex_rank_summarizer(
    text_parser.document,
    sentences_count=3
)
```

```python
# ============================================================
# Step 7: Display Summary
# ============================================================

# Display the selected important sentences.

print("LexRank Summary:\n")

for sentence in summary:
    print(sentence)
```
