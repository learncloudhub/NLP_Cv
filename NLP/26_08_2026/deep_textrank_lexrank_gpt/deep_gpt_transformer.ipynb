```python
# ============================================================
# GPT-2 Practical - Text Generation
# ============================================================

# Install if needed:
# !pip install transformers torch

from transformers import GPT2Tokenizer, GPT2LMHeadModel


# ============================================================
# Step 1: Load Tokenizer and GPT-2
# ============================================================

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

model = GPT2LMHeadModel.from_pretrained("gpt2")


# ============================================================
# Step 2: Create Input Text
# ============================================================

text = "Artificial intelligence is"


# ============================================================
# Step 3: Convert Text into Tokens
# ============================================================
# Convert text into GPT-2 tokens
inputs = tokenizer(
    text,
    return_tensors="pt"   # Return PyTorch tensor
)

print("Input Text:")
print(text)

print("\nToken IDs:")
print(inputs["input_ids"])


# Generate new text from the input
output = model.generate(
    inputs["input_ids"],    # Input token IDs
    max_length=30,          # Maximum number of tokens
    do_sample=True          # Randomly sample next tokens
)


# ============================================================
# Step 5: Convert Tokens into Text
# ============================================================

# Convert generated token IDs back into text
generated_text = tokenizer.decode(
    output[0],                 # Get the generated token IDs
    skip_special_tokens=True   # Remove special tokens
)

# ============================================================
# Step 6: Display Result
# ============================================================

print("\nGenerated Text:")
print(generated_text)
```
