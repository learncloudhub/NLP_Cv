```python
from nltk import CFG
from nltk.parse import ChartParser

# Define CFG grammar
grammar = CFG.fromstring("""
S -> NP VP
S -> NP VP CC NP VP

NP -> PN
NP -> Det N

VP -> V ADJ

PN -> 'i'
Det -> 'my'
N -> 'brother'
V -> 'am' | 'is'
ADJ -> 'good'
CC -> 'and'
""")

# Create parser
parser = ChartParser(grammar)

# Input sentence
sentence = "I am good and my brother is good"

# Tokenize
words = sentence.lower().split()

print("Words:", words)

# Parse
trees = list(parser.parse(words))

# Display tree
if trees:
    for tree in trees:
        print("\nParse Tree:")
        print(tree)
        tree.pretty_print()
else:
    print("No trees found.")
```
