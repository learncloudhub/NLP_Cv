```python
# bash
# pip install spacy
# python -m spacy download en_core_web_sm

import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Input sentence
sentence = "The boy eats rice."

# Process the sentence
doc = nlp(sentence)

# Display dependency information
print("Word\tPOS\tDependency\tHead")

for token in doc:
    print(
        token.text,
        "\t",
        token.pos_,
        "\t",
        token.dep_,
        "\t\t",
        token.head.text
    )
```
