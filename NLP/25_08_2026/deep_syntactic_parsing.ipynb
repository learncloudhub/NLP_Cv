```python
# ============================================================
# AI Diploma - NLP Practical 01
# Topic: Syntactic Parsing using NLTK
# ============================================================

import nltk


# ------------------------------------------------------------
# 1. Define the Context-Free Grammar (CFG)
# ------------------------------------------------------------

grammar = nltk.CFG.fromstring("""
    S -> NP VP

    NP -> Det N
    VP -> V NP

    Det -> 'the' | 'a'
    N -> 'student' | 'book' | 'teacher'
    V -> 'reads'
""")


# ------------------------------------------------------------
# 2. Create the Chart Parser
# ------------------------------------------------------------

parser = nltk.ChartParser(grammar)


# ------------------------------------------------------------
# 3. Define the sentence
# ------------------------------------------------------------

sentence = ['the', 'student', 'reads', 'a', 'book']


# ------------------------------------------------------------
# 4. Parse the sentence
# ------------------------------------------------------------

for tree in parser.parse(sentence):

    # Print the parse tree in bracket notation
    print(tree)

    # Display the parse tree visually
    tree.pretty_print()
```

```python
import nltk
from nltk import CFG
from nltk.parse import ChartParser
```

```python
grammar = CFG.fromstring("""
    S -> NP VP
    NP -> Pronoun
    NP -> Det N
    NP -> Det N PP
    VP -> V NP
    VP -> VP PP
    PP -> P NP

    Pronoun -> 'I'
    Det -> 'the' | 'a'
    N -> 'man' | 'telescope'
    V -> 'saw'
    P -> 'with'
""")
```

```python
# 3. Input sentence
parser = nltk.ChartParser(grammar)
sentence = "I saw the man with a telescope."

# Remove the full stop
sentence = sentence.replace(".","")
words = sentence.split()

# 4. Parse the sentence

print("Sentence:", sentence)
print("\nConstituency Parse Tree:")

# Generate all possible parse trees and store them in trees.
trees = list(parser.parse(words))
if trees:
    for tree in trees:
        print(tree)
        tree.pretty_print()
else:
    print("No trees found.")
```
